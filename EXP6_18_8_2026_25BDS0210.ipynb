```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("WineQT.csv")

print(df.head(3))

print(df.tail(3))
```

```python
print(df.dtypes)
```

```python
print(df.describe())
```

```python
print(df.isna().any())
```

```python
print(df.isna().sum())
```

```python
print("Missing values in each column:")
print(df.isnull().sum())

print("\nNumber of duplicate rows:")
print(df.duplicated().sum())
```

```python
df = df.drop_duplicates()
```

```python
mean = df["alcohol"].mean()
median = df["alcohol"].median()
mode = df["alcohol"].mode()

print("Mean =", mean)
print("Median =", median)
print("Mode =", mode)
```

```python
df["quality"].value_counts().sort_index().plot(kind="bar", figsize=(10,6))

plt.title("Number of Wines by Quality")
plt.ylabel("Number of Wines")
plt.xlabel("Quality")
plt.show()
```

```python
df["quality"].value_counts().sort_index().plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Number of Wines by Quality")
plt.xlabel("Wine Quality")
plt.ylabel("Number of Wines")
plt.show()
```

```python
sns.boxplot(x="alcohol", data=df)

plt.title("Boxplot of Alcohol")
plt.xlabel("Alcohol")
plt.show()
```

```python
sns.boxplot(x="alcohol", data=df)

plt.title("Boxplot of Alcohol")
plt.xlabel("Alcohol")
plt.show()
```

```python
plt.scatter(df["alcohol"], df["quality"])

plt.title("Scatter Plot for Alcohol vs Quality")
plt.xlabel("Alcohol")
plt.ylabel("Quality")
plt.show()
```

```python
sns.boxplot(x="quality", y="alcohol", data=df)

plt.title("Alcohol Distribution for Different Wine Quality Ratings")
plt.xlabel("Wine Quality")
plt.ylabel("Alcohol")
plt.show()
```

```python
sns.boxplot(x="quality", y="sulphates", data=df)

plt.title("Sulphates Distribution for Different Wine Quality Ratings")
plt.xlabel("Wine Quality")
plt.ylabel("Sulphates")
plt.show()
```

```python
sns.pairplot(
    df,
    vars=["alcohol", "volatile acidity", "sulphates", "quality"],
    kind="reg"
)

plt.show()
```

```python
sns.set(style="ticks", color_codes=True)

sns.pairplot(
    df,
    height=2,
    vars=["fixed acidity", "volatile acidity", "alcohol"],
    hue="quality"
)

plt.show()
```

```python
from scipy import stats

corr = stats.pearsonr(df["alcohol"], df["quality"])

print("p-value:\t", corr[1])
print("cor:\t\t", corr[0])
```

```python
correlation = df.corr(method="pearson", numeric_only=True)

print(correlation)
```

```python
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    xticklabels=correlation.columns,
    yticklabels=correlation.columns
)

plt.show()
```
