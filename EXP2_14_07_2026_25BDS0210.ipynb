```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("california_housing_test.csv")


# Display first 5 rows
print(df.head())

# Information about rows, columns, data types
df.info()

# Summary statistics
print(df.describe())

# Shape of dataset
print("Rows and Columns:", df.shape)

# Column names
print(df.columns)

# Check missing values
print(df.isnull().sum())


# Display selected columns
print(df[["median_income", "median_house_value"]])

# First 10 rows of selected columns
print(df[["longitude", "latitude"]].head(10))


# Houses with value greater than 300000
print(df[df["median_house_value"] > 300000])

# Houses with median income greater than 5
print(df[df["median_income"] > 5])

# Houses older than 40 years
print(df[df["housing_median_age"] > 40])

# Houses with more than 5000 rooms
print(df[df["total_rooms"] > 5000])


# Sort by house value
print(df.sort_values(by="median_house_value"))

# Sort by income (highest first)
print(df.sort_values(by="median_income", ascending=False))


# Mean house value
print("Average House Value:", df["median_house_value"].mean())

# Maximum income
print("Maximum Income:", df["median_income"].max())

# Minimum house value
print("Minimum House Value:", df["median_house_value"].min())

# Total population
print("Total Population:", df["population"].sum())

# Median house age
print("Median House Age:", df["housing_median_age"].median())

# Standard deviation of income
print("Income Std Dev:", df["median_income"].std())


# House with maximum value
print(df[df["median_house_value"] == df["median_house_value"].max()])

# House with minimum income
print(df[df["median_income"] == df["median_income"].min()])


# Number of houses with value above 200000
print((df["median_house_value"] > 200000).sum())

# Number of houses with income above average
print((df["median_income"] > df["median_income"].mean()).sum())


# Average rooms per household
df["rooms_per_household"] = df["total_rooms"] / df["households"]

# Average people per household
df["people_per_household"] = df["population"] / df["households"]

# Bedrooms per room ratio
df["bedroom_ratio"] = df["total_bedrooms"] / df["total_rooms"]

print(df.head())


# Correlation matrix
print(df.corr())

# Correlation with house value
print(df.corr()["median_house_value"].sort_values(ascending=False))

# Merge Example

df2 = pd.DataFrame({
    "longitude": [-122.05, -118.30, -117.81],
    "Area": ["Area A", "Area B", "Area C"]
})

merged = pd.merge(df, df2, on="longitude")
print(merged)

# Concatenation Example

df3 = pd.DataFrame({
    "longitude": [-120.00],
    "latitude": [36.00],
    "housing_median_age": [20],
    "total_rooms": [1500],
    "total_bedrooms": [300],
    "population": [700],
    "households": [250],
    "median_income": [4.5],
    "median_house_value": [180000],
    "rooms_per_household": [6],
    "people_per_household": [2.8],
    "bedroom_ratio": [0.2]
})

new_df = pd.concat([df, df3], ignore_index=True)

print(new_df.tail())
```
